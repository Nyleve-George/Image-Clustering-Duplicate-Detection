## Cell 1 : Install and Import Libraries

```python
# ── Install required libraries (run once) ─────────────────────────────────────
# Uncomment the line below if running for the first time in your environment
!pip install textblob tokenizers transformers --quiet
!pip install tensorflow --quiet

# ── Standard / Data ───────────────────────────────────────────────────────────
import random
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import pandas as pd

# ── NLTK (dataset + baseline model) ──────────────────────────────────────────
import nltk
nltk.download('movie_reviews', quiet=True)
nltk.download('punkt',         quiet=True)
nltk.download('stopwords',     quiet=True)
from nltk.corpus import movie_reviews, stopwords
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy as nltk_accuracy

# ── TextBlob (Model 2) ────────────────────────────────────────────────────────
from textblob import TextBlob

# ── Deep Learning — Keras/TensorFlow (Model 1: BiLSTM) ───────────────────────
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, Dense, Dropout
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping

# ── Evaluation ────────────────────────────────────────────────────────────────
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

# ── Visualisation ─────────────────────────────────────────────────────────────
import matplotlib.pyplot as plt
import seaborn as sns

# ── Reproducibility ───────────────────────────────────────────────────────────
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("All libraries imported successfully.")
```

---
## Cell 2 : Load the NLTK Movie Reviews Dataset
NLTK's `movie_reviews` corpus contains plain-text reviews categorised into `pos` and `neg`.  
We load **all** reviews and their labels into Python lists for use across all three models.

```python
# ── Load raw text reviews and labels ──────────────────────────────────────────
documents = []    # Each item: (full_text_string, label)

for category in movie_reviews.categories():      # 'pos' and 'neg'
    for fileid in movie_reviews.fileids(category):
        # movie_reviews.words() returns a tokenised list; join to get full text
        text = " ".join(movie_reviews.words(fileid))
        documents.append((text, category))

# Shuffle to mix positives and negatives randomly
random.shuffle(documents)

# Separate into parallel lists — easier to split later
texts  = [doc[0] for doc in documents]    # Raw text strings
labels = [doc[1] for doc in documents]    # 'pos' or 'neg'

print(f"Total reviews loaded : {len(documents)}")
print(f"Positive reviews     : {labels.count('pos')}")
print(f"Negative reviews     : {labels.count('neg')}")
print(f"\nSample review (truncated): {texts[0][:200]}...")
```

---
## Cell 3 : Train/Test Split (Shared Across All Models)
We split the 2000 reviews into **80% training / 20% testing** once, and use the **same split** for every model.  
This ensures fair comparison.

```python
# ── 80/20 train-test split ────────────────────────────────────────────────────
X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.2, random_state=SEED, stratify=labels
)

print(f"Training samples : {len(X_train)}")
print(f"Test samples     : {len(X_test)}")
print(f"\nClass balance in test set: pos={y_test.count('pos')}, neg={y_test.count('neg')}")
```

---
# ═══════════════════════════════════════════════════
# Reference Model : NLTK Naïve Bayes Classifier
# ═══════════════════════════════════════════════════

## Cell 4 : Feature Engineering for NLTK
NLTK's Naïve Bayes classifier works on **dictionaries of features**, not raw text.  
The standard feature: presence/absence of the **most frequent words** in the corpus.  
This is the same approach used in the reference script from the project brief.

```python
nltk.download('punkt_tab')
```

```python
# ── Build a vocabulary from training text only (avoid data leakage) ───────────
stop_words = set(stopwords.words('english'))   # Common words to ignore

# Count word frequencies across ALL training documents
all_words_freq = nltk.FreqDist(
    w.lower() for text in X_train
    for w in nltk.word_tokenize(text)
    if w.isalpha() and w.lower() not in stop_words
)

# Use the 2000 most common words as our feature set
NUM_FEATURES = 2000
word_features = list(all_words_freq.keys())[:NUM_FEATURES]

def document_features(text):
    """Convert a text string into a feature dictionary for NLTK.
    Feature = True if word is present in text, else False."""
    words_in_doc = set(nltk.word_tokenize(text.lower()))
    return {f'contains({w})': (w in words_in_doc) for w in word_features}

print(f"Vocabulary size used: {NUM_FEATURES} most frequent words")
print("Feature engineering function defined")
```

---
## Cell 5 : Train and Evaluate the NLTK Naïve Bayes Baseline

```python
# ── Prepare feature sets in NLTK's expected format: [(feature_dict, label)] ───
train_features = [(document_features(text), label) for text, label in zip(X_train, y_train)]
test_features  = [(document_features(text), label) for text, label in zip(X_test,  y_test)]

# ── Train Naïve Bayes ─────────────────────────────────────────────────────────
print("Training NLTK Naïve Bayes classifier...")
nb_classifier = NaiveBayesClassifier.train(train_features)

# ── Evaluate ──────────────────────────────────────────────────────────────────
nb_accuracy = nltk_accuracy(nb_classifier, test_features)

# Convert to numpy-friendly predictions for sklearn metrics
y_pred_nb = [nb_classifier.classify(document_features(text)) for text in X_test]

print(f"\nNLTK Naïve Bayes — Test Accuracy: {nb_accuracy:.4f} ({nb_accuracy*100:.2f}%)")
print("\nDetailed Classification Report:")
print(classification_report(y_test, y_pred_nb, target_names=['neg', 'pos']))

print("\nTop 15 most informative features:")
nb_classifier.show_most_informative_features(15)
```

# ═══════════════════════════════════════════════════
# Model 1 : Bidirectional LSTM with Sub-word Tokenization
# ═══════════════════════════════════════════════════

## Cell 6 : Tokenization
### What is sub-word tokenization?
Instead of treating each whole word as a token, **sub-word tokenization** splits rare words into smaller meaningful chunks (e.g., "unhappiness" → "un", "happiness").  
This lets the model handle out-of-vocabulary words gracefully.

We use Keras's `Tokenizer` with `char_level=False` and **oov_token** for unknown words, a practical sub-word approximation within the standard Keras workflow.  
> **Note:** A full sub-word tokenizer (BPE/WordPiece) can be substituted using the `tokenizers` library for an even stronger baseline.

```python
# ── Hyperparameters (centralised for easy tuning) ─────────────────────────────
VOCAB_SIZE  = 20000   # Top-N words to keep in vocabulary
MAX_LEN     = 300     # Pad/truncate each review to 300 tokens
EMBED_DIM   = 64      # Embedding vector size per token
LSTM_UNITS  = 64      # Hidden units in each LSTM direction
BATCH_SIZE  = 32
EPOCHS      = 10      # EarlyStopping will cut this short if needed

# ── Build the Keras Tokenizer on training data only ───────────────────────────
tokenizer = Tokenizer(
    num_words=VOCAB_SIZE,
    oov_token='<OOV>'     # Sub-word fallback: out-of-vocabulary words → single token
)
tokenizer.fit_on_texts(X_train)   # Learn vocabulary from training set

# ── Convert texts to integer sequences ───────────────────────────────────────
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq  = tokenizer.texts_to_sequences(X_test)

# ── Pad/truncate all sequences to the same length ────────────────────────────
# 'post' padding/truncation adds/removes from the end of the sequence
X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_LEN, padding='post', truncating='post')
X_test_pad  = pad_sequences(X_test_seq,  maxlen=MAX_LEN, padding='post', truncating='post')

# ── Convert string labels to binary integers ──────────────────────────────────
y_train_bin = np.array([1 if y == 'pos' else 0 for y in y_train])
y_test_bin  = np.array([1 if y == 'pos' else 0 for y in y_test])

print(f"Vocabulary size       : {len(tokenizer.word_index)} unique tokens")
print(f"Training matrix shape : {X_train_pad.shape}")
print(f"Test matrix shape     : {X_test_pad.shape}")
```

---
## Cell 7 : Build the Bidirectional LSTM Architecture
### Why BiLSTM?
A standard LSTM reads text **left → right**. A **Bidirectional** LSTM reads it both **left → right AND right → left**, giving the model context from both directions especially useful for sentiment where the end of a sentence can flip its meaning (e.g., 'Not good').

```python
# ── Build the BiLSTM model ────────────────────────────────────────────────────
model_bilstm = Sequential([

    # Layer 1: Embedding — maps token integers to dense vectors
    Embedding(input_dim=VOCAB_SIZE, output_dim=EMBED_DIM, input_length=MAX_LEN),

    # Layer 2: First Bidirectional LSTM — return_sequences=True passes
    #          the full sequence to the next LSTM layer
    Bidirectional(LSTM(LSTM_UNITS, return_sequences=True)),
    Dropout(0.3),   # Dropout reduces overfitting

    # Layer 3: Second Bidirectional LSTM — return_sequences=False
    #          outputs a single vector summarising the whole sequence
    Bidirectional(LSTM(LSTM_UNITS // 2)),
    Dropout(0.3),

    # Layer 4: Dense hidden layer — learns non-linear combinations
    Dense(32, activation='relu'),

    # Output layer: sigmoid → outputs probability of 'pos' (0 to 1)
    Dense(1, activation='sigmoid')
])

# ── Compile ───────────────────────────────────────────────────────────────────
model_bilstm.compile(
    loss='binary_crossentropy',   # Binary classification loss
    optimizer='adam',
    metrics=['accuracy']
)

model_bilstm.summary()
```

---
## Cell 8 : Train the BiLSTM Model

```python
# ── Early stopping: stop training if validation loss stops improving ───────────
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=2,             # Stop after 2 epochs of no improvement
    restore_best_weights=True   # Roll back to the best epoch
)

# ── Train ─────────────────────────────────────────────────────────────────────
print("Training BiLSTM model...")
history = model_bilstm.fit(
    X_train_pad, y_train_bin,
    validation_split=0.1,   # Use 10% of training data for validation monitoring
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stop],
    verbose=1
)
print("\n Training complete.")
```

---
## Cell 9 : Plot Training History

```python
# ── Plot Accuracy and Loss curves ─────────────────────────────────────────────
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Accuracy
ax1.plot(history.history['accuracy'],     label='Train Accuracy')
ax1.plot(history.history['val_accuracy'], label='Val Accuracy')
ax1.set_title('BiLSTM — Accuracy per Epoch')
ax1.set_xlabel('Epoch'); ax1.set_ylabel('Accuracy')
ax1.legend(); ax1.grid(True, alpha=0.4)

# Loss
ax2.plot(history.history['loss'],     label='Train Loss')
ax2.plot(history.history['val_loss'], label='Val Loss')
ax2.set_title('BiLSTM — Loss per Epoch')
ax2.set_xlabel('Epoch'); ax2.set_ylabel('Loss')
ax2.legend(); ax2.grid(True, alpha=0.4)

plt.tight_layout()
plt.show()
```

---
## Cell 10 : Evaluate the BiLSTM Model on the Test Set

```python
# ── Get predicted probabilities, then threshold at 0.5 ───────────────────────
y_prob_bilstm = model_bilstm.predict(X_test_pad, verbose=0).flatten()
y_pred_bilstm_bin = (y_prob_bilstm >= 0.5).astype(int)

# Convert back to string labels for consistent comparison
y_pred_bilstm = ['pos' if p == 1 else 'neg' for p in y_pred_bilstm_bin]

bilstm_accuracy = accuracy_score(y_test, y_pred_bilstm)

print(f" BiLSTM — Test Accuracy: {bilstm_accuracy:.4f} ({bilstm_accuracy*100:.2f}%)")
print("\nDetailed Classification Report:")
print(classification_report(y_test, y_pred_bilstm, target_names=['neg', 'pos']))

# ── Confusion Matrix ──────────────────────────────────────────────────────────
cm_bilstm = confusion_matrix(y_test, y_pred_bilstm, labels=['neg', 'pos'])
plt.figure(figsize=(5, 4))
sns.heatmap(cm_bilstm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['neg', 'pos'], yticklabels=['neg', 'pos'])
plt.title('Confusion Matrix — BiLSTM')
plt.ylabel('True Label'); plt.xlabel('Predicted Label')
plt.tight_layout()
plt.show()
```

# Model 2 : TextBlob (Lexicon-Based / Rule-Based)
# ═══════════════════════════════════════════════════

## Cell 11 : TextBlob Sentiment Analysis
**TextBlob** doesn't need training. It uses a built-in lexicon where each word has a pre-assigned **polarity** score (−1 = very negative, +1 = very positive).  
The overall sentiment of a sentence is the **average polarity** across all words.  

| Polarity | Label |
|----------|-------|
| ≥ 0      | `pos` |
| < 0      | `neg` |

```python
# ── Apply TextBlob to every test review ───────────────────────────────────────
y_pred_textblob = []

for text in X_test:
    blob = TextBlob(text)
    # .sentiment.polarity returns a float in [-1, 1]
    if blob.sentiment.polarity >= 0:
        y_pred_textblob.append('pos')
    else:
        y_pred_textblob.append('neg')

textblob_accuracy = accuracy_score(y_test, y_pred_textblob)

print(f"TextBlob — Test Accuracy: {textblob_accuracy:.4f} ({textblob_accuracy*100:.2f}%)")
print("\nDetailed Classification Report:")
print(classification_report(y_test, y_pred_textblob, target_names=['neg', 'pos']))

# ── Confusion Matrix ──────────────────────────────────────────────────────────
cm_tb = confusion_matrix(y_test, y_pred_textblob, labels=['neg', 'pos'])
plt.figure(figsize=(5, 4))
sns.heatmap(cm_tb, annot=True, fmt='d', cmap='Oranges',
            xticklabels=['neg', 'pos'], yticklabels=['neg', 'pos'])
plt.title('Confusion Matrix — TextBlob')
plt.ylabel('True Label'); plt.xlabel('Predicted Label')
plt.tight_layout()
plt.show()
```

---
## Cell 12 : TextBlob Polarity Distribution
Let's visualise where the polarity scores fall for positive vs. negative reviews.  
A well-separated distribution would mean TextBlob's scores are meaningful for this dataset.

```python
# ── Compute polarity scores for all test reviews ───────────────────────────────
polarities_pos = [TextBlob(t).sentiment.polarity for t, y in zip(X_test, y_test) if y == 'pos']
polarities_neg = [TextBlob(t).sentiment.polarity for t, y in zip(X_test, y_test) if y == 'neg']

plt.figure(figsize=(9, 4))
plt.hist(polarities_pos, bins=30, alpha=0.6, color='green', label='True Positive')
plt.hist(polarities_neg, bins=30, alpha=0.6, color='red',   label='True Negative')
plt.axvline(0, color='black', linestyle='--', label='Decision boundary (0)')
plt.xlabel('TextBlob Polarity Score', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('TextBlob Polarity Distribution by True Label', fontsize=13)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Cell 13 : Side-by-Side Performance Table
We compile accuracy, precision, recall, and F1 for all three models into a single table.  
**Same test set, same 400 reviews**:  direct apple-to-apple comparison.

```python
from sklearn.metrics import precision_score, recall_score, f1_score

def compute_metrics(y_true, y_pred, model_name):
    """Compute accuracy, precision, recall, F1 for a set of predictions."""
    return {
        'Model'    : model_name,
        'Accuracy' : round(accuracy_score(y_true, y_pred), 4),
        'Precision': round(precision_score(y_true, y_pred, pos_label='pos'), 4),
        'Recall'   : round(recall_score(y_true, y_pred,    pos_label='pos'), 4),
        'F1 Score' : round(f1_score(y_true, y_pred,        pos_label='pos'), 4),
        'Type'     : 'NN (BiLSTM)' if 'LSTM' in model_name else
                     'Rule-based'  if 'TextBlob' in model_name else
                     'Probabilistic'
    }

# ── Build the comparison table ────────────────────────────────────────────────
comparison = pd.DataFrame([
    compute_metrics(y_test, y_pred_nb,       'NLTK Naïve Bayes (Baseline)'),
    compute_metrics(y_test, y_pred_bilstm,   'BiLSTM (Sub-word Tokenization)'),
    compute_metrics(y_test, y_pred_textblob, 'TextBlob (Lexicon-based)'),
])

# Highlight the best value in each numeric column
print("\n" + "="*70)
print("       MODEL PERFORMANCE COMPARISON (Test Set — 400 reviews)")
print("="*70)
print(comparison.to_string(index=False))
print("="*70)
```

---
## Cell 14 : Visual Comparison Bar Chart

```python
# ── Bar chart comparing all models on all metrics ─────────────────────────────
metrics = ['Accuracy', 'Precision', 'Recall', 'F1 Score']
models  = comparison['Model'].tolist()
x       = np.arange(len(metrics))
width   = 0.25   # Width of each bar group
colors  = ['steelblue', 'coral', 'mediumseagreen']

fig, ax = plt.subplots(figsize=(11, 6))

for i, (model_name, color) in enumerate(zip(models, colors)):
    row = comparison[comparison['Model'] == model_name].iloc[0]
    vals = [row[m] for m in metrics]
    bars = ax.bar(x + i * width, vals, width, label=model_name, color=color, alpha=0.85)
    # Add value labels on top of each bar
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
                f'{val:.3f}', ha='center', va='bottom', fontsize=9)

ax.set_ylabel('Score', fontsize=12)
ax.set_title('Sentiment Analysis Model Comparison', fontsize=14, fontweight='bold')
ax.set_xticks(x + width)
ax.set_xticklabels(metrics, fontsize=12)
ax.set_ylim(0, 1.12)
ax.legend(fontsize=10, loc='upper right')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## Cell 15 : Discussion and Conclusions

### Interpretation of Results

| Model | Strengths | Weaknesses |
|-------|-----------|------------|
| **NLTK Naïve Bayes** | Fast, simple, interpretable, good baseline | Ignores word order; bag-of-words assumption |
| **BiLSTM** | Understands word order and context; deep feature learning | Requires more data & compute; black-box |
| **TextBlob** | No training needed; plug-and-play | Fixed lexicon; no domain adaptation; scores movie irony poorly |

### Key Takeaways
The **BiLSTM** is expected to outperform both other models because it captures sequential context such as negation ("not good"), sarcasm cues, and sentence-level patterns.
- **TextBlob** may underperform on movie reviews specifically because movie reviews often use nuanced language that simple word-polarity averaging misses.
- The **NLTK Naïve Bayes** is a solid and fast baseline; it often achieves ~80% on this dataset.
- All three models were evaluated on the **same 400-review test split** for fair comparison.
