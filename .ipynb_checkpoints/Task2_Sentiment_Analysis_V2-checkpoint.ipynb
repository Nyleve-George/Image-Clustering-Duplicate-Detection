# Task 2: Custom Sentiment Analysis — Train, Evaluate and Compare

## Overview

This notebook trains and evaluates three sentiment analysis models on the NLTK Movie Reviews
dataset, which contains 2000 labelled movie reviews: 1000 positive and 1000 negative.

The reference implementation provided in the project brief uses an NLTK Naive Bayes classifier.
This notebook reproduces that baseline and then introduces two custom models:

- Model 1: A Bidirectional LSTM neural network using Byte Pair Encoding (BPE) sub-word tokenization.
- Model 2: A TextBlob lexicon-based classifier that requires no training.

All three models are evaluated on the same held-out test set and compared in a performance table.

## Dataset

| Property | Value |
|----------|-------|
| Source | nltk.corpus.movie_reviews |
| Total reviews | 2000 |
| Positive reviews | 1000 |
| Negative reviews | 1000 |
| Task | Binary sentiment classification |


## Section 1: Install and Import Libraries

All packages are installed first. The tokenizers library from Hugging Face provides the true
Byte Pair Encoding (BPE) tokenizer used for Model 1. TensorFlow and Keras handle the neural
network architecture. NLTK provides the dataset and the Naive Bayes baseline.


```python
# Run this cell first. The quiet flag suppresses installation output.
import subprocess
subprocess.run(['pip', 'install', 'textblob', 'tokenizers', '--quiet'], check=False)
subprocess.run(['pip', 'install', 'tensorflow', '--quiet'], check=False)
subprocess.run(['pip', 'install', 'nltk', '--quiet'], check=False)
print("Installation complete.")
```

```python
import random
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# NLTK components
import nltk
nltk.download('movie_reviews', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
from nltk.corpus import movie_reviews, stopwords
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy as nltk_accuracy

# TextBlob for Model 2
from textblob import TextBlob

# Hugging Face tokenizers for BPE sub-word tokenization (Model 1)
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# TensorFlow and Keras for the BiLSTM neural network (Model 1)
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, Dense, Dropout
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping

# Evaluation metrics
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report, accuracy_score, confusion_matrix,
    precision_score, recall_score, f1_score
)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("All libraries imported successfully.")
```

## Section 2: Load the NLTK Movie Reviews Dataset

The NLTK movie_reviews corpus is stored as a collection of plain text files organised into two
category folders: 'pos' and 'neg'. Each file is one movie review.

The words from each file are joined into a single string. The resulting list of (text, label)
pairs is shuffled to ensure a random mix of positive and negative examples before splitting.


```python
documents = []

for category in movie_reviews.categories():
    for fileid in movie_reviews.fileids(category):
        text = " ".join(movie_reviews.words(fileid))
        documents.append((text, category))

random.shuffle(documents)

texts  = [doc[0] for doc in documents]
labels = [doc[1] for doc in documents]

print(f"Total reviews loaded : {len(documents)}")
print(f"Positive reviews     : {labels.count('pos')}")
print(f"Negative reviews     : {labels.count('neg')}")
print(f"\nSample review (first 250 characters):")
print(texts[0][:250] + "...")
```

## Section 3: Train and Test Split

The 2000 reviews are split 80% for training and 20% for testing. The split is stratified so that
both the training and test sets contain exactly equal proportions of positive and negative reviews.

The same split is shared across all three models. This is essential for a fair comparison: if each
model were evaluated on a different subset, differences in results could be due to the composition
of the test set rather than the model's actual capability.

The random_state is fixed so results are reproducible across notebook runs.


```python
X_train, X_test, y_train, y_test = train_test_split(
    texts, labels,
    test_size=0.2,
    random_state=SEED,
    stratify=labels
)

print(f"Training samples : {len(X_train)}")
print(f"Test samples     : {len(X_test)}")
print(f"\nClass balance in training: pos = {y_train.count('pos')}, neg = {y_train.count('neg')}")
print(f"Class balance in test    : pos = {y_test.count('pos')}, neg = {y_test.count('neg')}")
```

## Section 4: Reference Model — NLTK Naive Bayes Classifier

This is the baseline model from the project reference script. It uses a bag-of-words
approach: each review is represented as a set of binary features indicating which of the top
2000 most frequent words appear in the review.

The vocabulary is built from the training set only, which prevents data leakage. Words that
appear in the test set but not in the training vocabulary are simply ignored, which is the
correct procedure for a fair evaluation.

The Naive Bayes classifier assumes that each word appears independently of the others given
the class label. Despite this strong assumption, it achieves solid performance on this dataset
because sentiment is strongly correlated with individual word presence.


```python
stop_words = set(stopwords.words('english'))

all_words_freq = nltk.FreqDist(
    w.lower()
    for text in X_train
    for w in nltk.word_tokenize(text)
    if w.isalpha() and w.lower() not in stop_words
)

NUM_FEATURES = 2000
word_features = list(all_words_freq.keys())[:NUM_FEATURES]

def document_features(text):
    words_in_doc = set(nltk.word_tokenize(text.lower()))
    return {f'contains({w})': (w in words_in_doc) for w in word_features}

print(f"Vocabulary built from training set only: {NUM_FEATURES} most frequent non-stopword words.")

train_features = [(document_features(text), label) for text, label in zip(X_train, y_train)]
test_features  = [(document_features(text), label) for text, label in zip(X_test,  y_test)]

print("Training NLTK Naive Bayes classifier...")
nb_classifier = NaiveBayesClassifier.train(train_features)

nb_accuracy = nltk_accuracy(nb_classifier, test_features)
y_pred_nb = [nb_classifier.classify(document_features(text)) for text in X_test]

print(f"\nNLTK Naive Bayes Test Accuracy: {nb_accuracy:.4f}  ({nb_accuracy * 100:.2f}%)")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_nb, target_names=['neg', 'pos']))
print("\nTop 15 most informative features:")
nb_classifier.show_most_informative_features(15)
```

## Section 5: Confusion Matrix for the Naive Bayes Baseline

A confusion matrix shows exactly how many reviews were correctly and incorrectly classified.
The four cells represent true positives, true negatives, false positives, and false negatives.
This provides more detail than accuracy alone, particularly for understanding which direction
errors tend to go.


```python
cm_nb = confusion_matrix(y_test, y_pred_nb, labels=['neg', 'pos'])

plt.figure(figsize=(5, 4))
sns.heatmap(cm_nb, annot=True, fmt='d', cmap='Greens',
            xticklabels=['Predicted Neg', 'Predicted Pos'],
            yticklabels=['True Neg', 'True Pos'])
plt.title('Confusion Matrix: NLTK Naive Bayes (Baseline)', fontsize=12)
plt.tight_layout()
plt.show()
```

## Section 6: Model 1 — Byte Pair Encoding (BPE) Sub-word Tokenization

### What is sub-word tokenization?

Standard word tokenization treats each unique word as an atomic token. This causes two problems.
First, rare words and domain-specific vocabulary (like character names or film terminology) are
mapped to an unknown token and their meaning is lost. Second, the vocabulary can grow very large
because each inflected form of a word counts separately.

Byte Pair Encoding (BPE) is a sub-word tokenization algorithm that solves both problems. It starts
with individual characters as the base vocabulary and repeatedly merges the most frequent pair of
adjacent tokens into a new token. After training, common words are represented as single tokens,
while rare words are split into smaller recognisable pieces.

For example, a word like "cinematography" might be split into "cinema", "togr", "aphy" if those
sub-strings appear frequently across the training corpus. The model can still process and learn
from the word rather than ignoring it entirely.

BPE was originally introduced by Sennrich et al. (2016) and is the tokenization method used by
GPT models. The implementation here uses the Hugging Face tokenizers library, which provides a
fast and well-tested BPE trainer.

Reference: Sennrich, R., Haddow, B., and Birch, A. (2016). Neural Machine Translation of Rare
Words with Subword Units. Proceedings of ACL 2016.


```python
VOCAB_SIZE = 10000
MAX_LEN    = 300
EMBED_DIM  = 64
LSTM_UNITS = 64
BATCH_SIZE = 32
EPOCHS     = 15

bpe_tokenizer = Tokenizer(BPE(unk_token="[UNK]"))

# ByteLevel pre-tokenizer handles punctuation and capitalisation correctly
# It splits on whitespace and punctuation boundaries and lowercases everything
from tokenizers.pre_tokenizers import ByteLevel
from tokenizers.decoders import ByteLevel as ByteLevelDecoder

bpe_tokenizer.pre_tokenizer = ByteLevel(add_prefix_space=False)
bpe_tokenizer.decoder = ByteLevelDecoder()

trainer = BpeTrainer(
    vocab_size=VOCAB_SIZE,
    special_tokens=["[UNK]", "[PAD]"],
    min_frequency=2
)

bpe_tokenizer.train_from_iterator(
    (t.lower() for t in X_train), trainer=trainer
)

actual_vocab_size = bpe_tokenizer.get_vocab_size()
PAD_ID = bpe_tokenizer.token_to_id("[PAD]")

print(f"BPE tokenizer trained on training set only.")
print(f"Requested vocabulary size : {VOCAB_SIZE}")
print(f"Actual vocabulary size    : {actual_vocab_size}")
print(f"PAD token id              : {PAD_ID}")

examples = [
    "this movie was cinematographically outstanding.",
    "an unbelievably disappointing and forgettable film.",
    "masterfully crafted with unforgettable performances."
]

print("\nSub-word tokenization examples:")
for ex in examples:
    encoded = bpe_tokenizer.encode(ex.lower())
    print(f"  Input  : {ex}")
    print(f"  Tokens : {encoded.tokens}")
    print()
```

## Section 7: Encode and Pad Sequences for the BiLSTM

The BPE tokenizer converts each review from a string into a list of integer token IDs. These
sequences are then padded or truncated to a fixed length of MAX_LEN tokens so they can be
processed in batches by the neural network.

Post-padding (adding zeros at the end) and post-truncation (cutting from the end) are used.
This preserves the beginning of each review, where sentiment-setting language typically appears.

Labels are converted from the strings 'pos' and 'neg' to binary integers 1 and 0.


```python
def encode_and_pad(texts, tokenizer, max_len, pad_id):
    texts = [t.lower() for t in texts]
    sequences = [enc.ids for enc in tokenizer.encode_batch(texts)]
    return pad_sequences(sequences, maxlen=max_len, padding='post',
                         truncating='post', value=pad_id)

X_train_pad = encode_and_pad(X_train, bpe_tokenizer, MAX_LEN, PAD_ID)
X_test_pad  = encode_and_pad(X_test,  bpe_tokenizer, MAX_LEN, PAD_ID)

y_train_bin = np.array([1 if y == 'pos' else 0 for y in y_train])
y_test_bin  = np.array([1 if y == 'pos' else 0 for y in y_test])

print(f"Training matrix shape : {X_train_pad.shape}")
print(f"Test matrix shape     : {X_test_pad.shape}")
print(f"\nSample sequence (first 20 token IDs): {X_train_pad[0][:20]}")
print(f"Sample label: {y_train_bin[0]}  ({'pos' if y_train_bin[0] == 1 else 'neg'})")
```

## Section 8: Model 1 — Bidirectional LSTM Architecture

### Why Bidirectional LSTM?

A standard LSTM reads text from left to right, building a hidden state that captures context
from earlier words. This means the model can use context like "This film was not..." when it
reaches the word "good", because it has already processed "not".

However, a left-to-right LSTM cannot use context that comes after a word. Sentiment expressions
sometimes depend on words that come later in the sentence. A Bidirectional LSTM runs two LSTMs
in parallel: one left to right and one right to left. The outputs are concatenated at each
position, giving the model access to both past and future context simultaneously.

This is particularly valuable for movie review sentiment, where a reviewer may build up context
before delivering a judgement, or use ironic framing that requires reading the whole sentence.

### Architecture

- Embedding layer: maps each token ID to a dense 64-dimensional vector. These vectors are learned
  during training rather than being fixed, so the model learns representations suited to sentiment.
- First Bidirectional LSTM (64 units per direction): returns the full sequence of hidden states.
- Dropout 0.4: randomly zeros out 40% of activations during training to reduce overfitting.
- Second Bidirectional LSTM (32 units per direction): produces a single summary vector.
- Dropout 0.4: applied again for regularisation.
- Dense layer (64 units, ReLU): learns non-linear combinations of the LSTM output.
- Output layer (1 unit, sigmoid): outputs a probability between 0 and 1. Values above 0.5
  are classified as positive sentiment.


## Section 8 & 9: Train the BiLSTM Model

Early stopping monitors the validation loss during training. If the validation loss does not
improve for three consecutive epochs, training stops and the weights from the best epoch are
restored. This prevents the model from overfitting to the training data by training for too
many epochs.

A validation split of 10% is taken from the training data to monitor generalisation during
training. This data is not part of the final test set.


```python
model_bilstm = Sequential([
    Embedding(input_dim=actual_vocab_size, output_dim=EMBED_DIM, input_length=MAX_LEN,
              mask_zero=True),
    Bidirectional(LSTM(LSTM_UNITS, return_sequences=True)),
    Dropout(0.3),
    Bidirectional(LSTM(LSTM_UNITS // 2, return_sequences=False)),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid')
])

model_bilstm.compile(
    loss='binary_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    metrics=['accuracy']
)

model_bilstm.build(input_shape=(None, MAX_LEN))
model_bilstm.summary()

early_stop = EarlyStopping(
    monitor='val_accuracy',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_pad, y_train_bin,
    test_size=0.1,
    random_state=SEED,
    stratify=y_train_bin
)

print("Training BiLSTM model with BPE sub-word tokenization...")
history = model_bilstm.fit(
    X_tr, y_tr,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stop],
    verbose=1
)
print("\nTraining complete.")
```

## Section 10: Training History Plots

Plotting accuracy and loss over training epochs provides insight into whether the model learned
well and whether overfitting occurred. Ideally, both training and validation metrics should
improve together and then stabilise. If training accuracy continues to improve while validation
accuracy drops, this indicates overfitting.


```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(history.history['accuracy'], label='Train Accuracy', color='steelblue')
ax1.plot(history.history['val_accuracy'], label='Validation Accuracy', color='tomato')
ax1.set_title('BiLSTM Training Accuracy per Epoch', fontsize=12)
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Accuracy')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(history.history['loss'], label='Train Loss', color='steelblue')
ax2.plot(history.history['val_loss'], label='Validation Loss', color='tomato')
ax2.set_title('BiLSTM Training Loss per Epoch', fontsize=12)
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Loss')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Section 11: Evaluate the BiLSTM on the Test Set

The trained model produces a probability between 0 and 1 for each review. A threshold of 0.5
is applied: probabilities at or above 0.5 are classified as positive and below as negative.

The classification report shows precision, recall, and F1 score for each class separately.
This is more informative than accuracy alone because it reveals whether the model has a bias
toward one class.


```python
y_prob_bilstm = model_bilstm.predict(X_test_pad, verbose=0).flatten()
y_pred_bilstm_bin = (y_prob_bilstm >= 0.5).astype(int)
y_pred_bilstm = ['pos' if p == 1 else 'neg' for p in y_pred_bilstm_bin]

bilstm_accuracy = accuracy_score(y_test, y_pred_bilstm)
print(f"BiLSTM Test Accuracy: {bilstm_accuracy:.4f}  ({bilstm_accuracy * 100:.2f}%)")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_bilstm, target_names=['neg', 'pos']))

cm_bilstm = confusion_matrix(y_test, y_pred_bilstm, labels=['neg', 'pos'])
plt.figure(figsize=(5, 4))
sns.heatmap(cm_bilstm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicted Neg', 'Predicted Pos'],
            yticklabels=['True Neg', 'True Pos'])
plt.title('Confusion Matrix: BiLSTM with BPE Tokenization', fontsize=12)
plt.tight_layout()
plt.show()

# Distribution of predicted probabilities
plt.figure(figsize=(8, 4))
plt.hist([y_prob_bilstm[i] for i in range(len(y_test)) if y_test[i] == 'pos'],
         bins=30, alpha=0.6, color='steelblue', label='True Positive')
plt.hist([y_prob_bilstm[i] for i in range(len(y_test)) if y_test[i] == 'neg'],
         bins=30, alpha=0.6, color='tomato', label='True Negative')
plt.axvline(0.5, color='black', linestyle='--', label='Decision boundary (0.5)')
plt.xlabel('Predicted Probability of Positive Sentiment', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
plt.title('BiLSTM Predicted Probability Distribution', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Section 12: Model 2 — TextBlob Lexicon-Based Sentiment Analysis

TextBlob uses a pre-built lexicon in which each word has a fixed polarity score between -1 and +1.
The overall sentiment of a review is calculated as the average polarity across all the words it
contains. No training is required because the lexicon is fixed.

A polarity score at or above 0 is classified as positive. A score below 0 is classified as negative.

The main advantage of TextBlob is speed and simplicity: it can classify any text instantly without
a training phase. The main limitation is that its lexicon is domain-general and cannot adapt to the
specific vocabulary and tone of movie reviews. Irony, sarcasm, and negation are also difficult for
a simple averaging approach to handle correctly.

TextBlob is applied to the same 400-review test set used by the other models.


```python
# Calibrate threshold from training set so we are not leaking test information
train_polarities = [TextBlob(t).sentiment.polarity for t in X_train]
pos_train_pol = [p for p, y in zip(train_polarities, y_train) if y == 'pos']
neg_train_pol = [p for p, y in zip(train_polarities, y_train) if y == 'neg']
THRESHOLD = (np.mean(pos_train_pol) + np.mean(neg_train_pol)) / 2
print(f"Calibrated TextBlob threshold: {THRESHOLD:.4f}")
print(f"Mean polarity of positive training reviews: {np.mean(pos_train_pol):.4f}")
print(f"Mean polarity of negative training reviews: {np.mean(neg_train_pol):.4f}")

y_pred_textblob = []
for text in X_test:
    blob = TextBlob(text)
    if blob.sentiment.polarity >= THRESHOLD:
        y_pred_textblob.append('pos')
    else:
        y_pred_textblob.append('neg')

textblob_accuracy = accuracy_score(y_test, y_pred_textblob)
print(f"\nTextBlob Test Accuracy: {textblob_accuracy:.4f}  ({textblob_accuracy * 100:.2f}%)")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_textblob, target_names=['neg', 'pos']))

cm_tb = confusion_matrix(y_test, y_pred_textblob, labels=['neg', 'pos'])
plt.figure(figsize=(5, 4))
sns.heatmap(cm_tb, annot=True, fmt='d', cmap='Oranges',
            xticklabels=['Predicted Neg', 'Predicted Pos'],
            yticklabels=['True Neg', 'True Pos'])
plt.title('Confusion Matrix: TextBlob (Lexicon-Based)', fontsize=12)
plt.tight_layout()
plt.show()
```

## Section 13: TextBlob Polarity Distribution

Plotting the polarity score distributions for true positive and true negative reviews reveals how
well the TextBlob scores separate the two classes. If the distributions overlap heavily near zero,
it means TextBlob struggles to distinguish sentiment on this dataset. A clean separation would
indicate that the lexicon polarity scores correlate well with actual movie sentiment labels.


```python
polarities_pos = [
    TextBlob(t).sentiment.polarity
    for t, y in zip(X_test, y_test) if y == 'pos'
]
polarities_neg = [
    TextBlob(t).sentiment.polarity
    for t, y in zip(X_test, y_test) if y == 'neg'
]

plt.figure(figsize=(9, 4))
plt.hist(polarities_pos, bins=30, alpha=0.6, color='steelblue', label='True Positive')
plt.hist(polarities_neg, bins=30, alpha=0.6, color='tomato', label='True Negative')
plt.axvline(0, color='black', linestyle='--', label='Decision boundary (polarity = 0)')
plt.xlabel('TextBlob Polarity Score', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
plt.title('TextBlob Polarity Score Distribution by True Label', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Average polarity for true positive reviews: {np.mean(polarities_pos):.4f}")
print(f"Average polarity for true negative reviews: {np.mean(polarities_neg):.4f}")
print(f"Overlap near zero makes classification difficult for TextBlob.")
```

## Section 14: Final Comparison of All Three Models

All three models are evaluated on the same 400-review test set. Accuracy alone can be misleading,
so precision, recall, and F1 score are also reported for each model.

- Precision: of all reviews predicted as positive, what fraction were actually positive?
- Recall: of all reviews that were actually positive, what fraction were correctly predicted?
- F1 Score: the harmonic mean of precision and recall. This is the most balanced single metric
  for binary classification when both false positives and false negatives matter.

A comparison table is printed and then visualised as a grouped bar chart.


```python
def compute_metrics(y_true, y_pred, model_name, model_type):
    return {
        'Model': model_name,
        'Type': model_type,
        'Accuracy':  round(accuracy_score(y_true, y_pred), 4),
        'Precision': round(precision_score(y_true, y_pred, pos_label='pos'), 4),
        'Recall':    round(recall_score(y_true, y_pred, pos_label='pos'), 4),
        'F1 Score':  round(f1_score(y_true, y_pred, pos_label='pos'), 4),
    }

comparison = pd.DataFrame([
    compute_metrics(y_test, y_pred_nb,       'NLTK Naive Bayes (Baseline)', 'Probabilistic'),
    compute_metrics(y_test, y_pred_bilstm,   'BiLSTM with BPE Tokenization', 'Neural Network'),
    compute_metrics(y_test, y_pred_textblob, 'TextBlob (Lexicon-Based)', 'Rule-Based'),
])

print("=" * 72)
print("   MODEL PERFORMANCE COMPARISON  (Test Set: 400 reviews)")
print("=" * 72)
print(comparison.to_string(index=False))
print("=" * 72)
```

```python
metrics = ['Accuracy', 'Precision', 'Recall', 'F1 Score']
model_names = comparison['Model'].tolist()
x = np.arange(len(metrics))
width = 0.25
colors = ['steelblue', 'tomato', 'darkorange']

fig, ax = plt.subplots(figsize=(12, 6))

for i, (name, color) in enumerate(zip(model_names, colors)):
    row = comparison[comparison['Model'] == name].iloc[0]
    vals = [row[m] for m in metrics]
    bars = ax.bar(x + i * width, vals, width, label=name, color=color, alpha=0.85)
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width() / 2,
                bar.get_height() + 0.005,
                f'{val:.3f}', ha='center', va='bottom', fontsize=9)

ax.set_ylabel('Score', fontsize=12)
ax.set_title('Sentiment Analysis Model Comparison (Test Set)', fontsize=13, fontweight='bold')
ax.set_xticks(x + width)
ax.set_xticklabels(metrics, fontsize=12)
ax.set_ylim(0, 1.15)
ax.legend(fontsize=10, loc='upper right')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

# Section 15: Discussion and Conclusions

### How each model worksThe NLTK Naive Bayes classifier treats each review as a bag of words and uses Bayes' theorem
to classify sentiment. It ignores word order but performs well because sentiment is often carried
by individual high-signal words like "awful", "waste", and "beautifully".

The Bidirectional LSTM reads the review as a sequence of BPE sub-word tokens, processing it
in both directions to capture context, negation, and sentence-level sentiment patterns.

TextBlob assigns polarity scores from a fixed lexicon and averages them across the review.
A calibrated threshold from the training set was used instead of the default zero, improving
accuracy from 58% to 75%.

### Observed results

| Model | Accuracy | Precision | Recall | F1 Score |
|-------|----------|-----------|--------|----------|
| NLTK Naive Bayes (Baseline) | 76.75% | 0.8057 | 0.7050 | 0.7520 |
| BiLSTM with BPE Tokenization | 61.75% | 0.5992 | 0.7100 | 0.6499 |
| TextBlob (Lexicon-Based) | 75.00% | 0.7500 | 0.7500 | 0.7500 |

### Why results differ from the expected order

The results differ from the expected performance order. The BiLSTM was expected to outperform
both other models but scored the lowest at 61.75%. This outcome is explained by the size of
the dataset.

The NLTK movie reviews corpus contains 2000 reviews, of which only 1440 are available for
training after the 80/20 split. Bidirectional LSTMs are data-hungry models that typically
require tens of thousands of labelled examples to learn generalisable sequential patterns.
With 1440 samples, the model overfits rapidly. Training accuracy reached 99% by epoch 5 while
validation accuracy plateaued at 63%, indicating the model memorised the training set rather
than learning transferable sentiment representations. Early stopping restored the best weights
from epoch 4, yielding a test accuracy of 61.75%.

Naive Bayes performs well precisely because it does not need sequential context. Sentiment in
movie reviews is strongly correlated with individual word presence, which is exactly what the
bag-of-words model captures. The BiLSTM would be expected to outperform Naive Bayes on a
larger corpus such as the Stanford Large Movie Review Dataset which contains 25,000 training
samples.

### Strengths and limitations

| Model | Key Strength | Key Limitation |
|-------|-------------|----------------|
| NLTK Naive Bayes | Fast, interpretable, strong on small data | Ignores word order and context |
| BiLSTM + BPE | Captures sequence, context, and sub-word meaning | Requires large data to generalise |
| TextBlob | Zero training, instant deployment, calibratable |Naive Bayes proved the strongest performer on this small corpus. TextBlob benefited from
threshold calibration. The BiLSTM was limited by dataset size rather than model quality. close the performance gap significantly with more training data.
