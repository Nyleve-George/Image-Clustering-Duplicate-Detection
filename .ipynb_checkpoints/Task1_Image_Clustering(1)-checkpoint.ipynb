# Task 1: Image Clustering and Duplicate Detection
**Final Project — 60 Marks**

## Objective
We have 300+ competitor social media images stored in a common folder. Our goal is to:
- **(a)** Use pre-extracted image vectors (InceptionV3 embeddings) to cluster visually similar images
- **(b)** Build a dictionary mapping each cluster label → list of image filenames
- **(c)** Move images into cluster folders and display 5 random samples per cluster
- **(d)** Identify duplicate images using Cosine Distance

---
### Why image vectors?
Raw pixel data is too high-dimensional and noisy for clustering. Instead, we use **InceptionV3 embeddings** — a 2048-dimensional feature vector per image extracted from a pretrained deep learning model. These vectors capture high-level visual semantics (shapes, textures, objects), making them ideal for similarity-based clustering.


---

## Cell 1: Import Libraries

We import the required libraries for data handling, clustering, evaluation, dimensionality reduction, distance computation, and visualisation. Each import is commented to explain its role.

```python
# ── Standard library ──────────────────────────────────────────────────────────
import os                          # File/folder operations
import shutil                      # Moving files between directories
import random                      # Picking random images for display
import warnings
warnings.filterwarnings('ignore')  # Keep output clean

# ── Data handling ─────────────────────────────────────────────────────────────
import numpy as np
import pandas as pd

# ── Clustering algorithms ─────────────────────────────────────────────────────
from sklearn.cluster import KMeans, AgglomerativeClustering

# ── Cluster evaluation metrics ────────────────────────────────────────────────
from sklearn.metrics import silhouette_score

# ── Dimensionality reduction (for visualisation only) ─────────────────────────
from sklearn.decomposition import PCA

# ── Distance / similarity ─────────────────────────────────────────────────────
from sklearn.metrics.pairwise import cosine_distances, cosine_similarity

# ── Visualisation ─────────────────────────────────────────────────────────────
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

print("All libraries imported successfully.")
```

---
## Cell 2 : Load the Image Vectors (Embeddings)
The CSV file `image_vectors_InceptionV3.csv` was used for this project.  
Each **row = one image**, each **column = one feature dimension** (2048 total).  
The first column is the image filename.

```python
# ── Load the CSV containing one embedding vector per image ────────────────────
df_raw = pd.read_csv('image_vectors_InceptionV3.csv')

# Skip the first 2 metadata rows (type row + meta row)
df = df_raw.iloc[2:].reset_index(drop=True)

filenames = df['image'].tolist()

#To get the features
feature_cols = [c for c in df.columns if c.startswith('n')]
X = df[feature_cols].astype(float).values

# The index holds the image filenames; the 2048 columns are the feature values
print(f"Dataset shape : {X.shape}")
print(f"Number of images  : {X.shape[0]}")
print(f"Embedding dimensions: {X.shape[1]}")
print("\nFirst 5 filenames:")
print(filenames[:5])

```

---
## Cell 3 : Dimensionality Reduction with PCA
2048 dimensions is very large. We use **PCA** to compress vectors to 50 components.  
This:
- Speeds up clustering significantly
- Removes noise while retaining ~95%+ of variance
- Does **not** change the clustering logic, only makes it practical

```python
# ── Reduce 2048-D → 50-D using PCA ───────────────────────────────────────────
pca = PCA(n_components=50, random_state=42)
X_pca = pca.fit_transform(X)

# How much of the original information is retained?
explained = pca.explained_variance_ratio_.sum() * 100
print(f"Variance retained after PCA (50 components): {explained:.2f}%")
print(f"Reduced shape: {X_pca.shape}  (images × components)")
```

---


### Cell 4: Elbow Method for K-Means

We run K-Means for k = 2 to 15 and record the inertia (within-cluster sum of squares) at each step. The point where inertia stops decreasing sharply (the "elbow") indicates a suitable k.

Reference: https://builtin.com/data-science/elbow-method

```python
# ── Elbow Method ──────────────────────────────────────────────────────────────
inertia_values = []        # Stores within-cluster sum of squares
k_range = range(2, 16)     # Test k = 2 through 15

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_pca)
    inertia_values.append(km.inertia_)
    print(f"  k={k:2d}  inertia={km.inertia_:,.0f}")

# ── Plot the Elbow Curve ───────────────────────────────────────────────────────
plt.figure(figsize=(9, 4))
plt.plot(list(k_range), inertia_values, 'bo-', linewidth=2, markersize=7)
plt.xlabel('Number of Clusters (k)', fontsize=13)
plt.ylabel('Inertia (Within-Cluster SSE)', fontsize=13)
plt.title('Elbow Method — K-Means', fontsize=15)
plt.xticks(list(k_range))
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()

print("\nLook for the 'elbow' — the k where the curve bends and flattens.")
```

## Cell 5 (a) : Silhouette Score for K-Means
**Silhouette score** measures how well each point fits its own cluster vs. the nearest neighbouring cluster.  
- Range: −1 to +1; **higher is better**
- This gives a more reliable indication of the correct k than the elbow alone.



```python
# ── Silhouette Score for K-Means ──────────────────────────────────────────────
kmeans_sil_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    kmeans_sil_scores.append(score)
    print(f"  k={k:2d}  silhouette score={score:.4f}")

best_k_kmeans = list(k_range)[np.argmax(kmeans_sil_scores)]
print(f"\nBest k for K-Means (highest silhouette): k = {best_k_kmeans}")

# ── Plot Silhouette Scores ─────────────────────────────────────────────────────
plt.figure(figsize=(9, 4))
plt.plot(list(k_range), kmeans_sil_scores, 'rs-', linewidth=2, markersize=7)
plt.xlabel('Number of Clusters (k)', fontsize=13)
plt.ylabel('Silhouette Score', fontsize=13)
plt.title('Silhouette Score — K-Means', fontsize=15)
plt.xticks(list(k_range))
plt.axvline(best_k_kmeans, color='green', linestyle='--', label=f'Best k={best_k_kmeans}')
plt.legend()
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()
```

---
## Cell 6 (a) : Second Clustering Method: Agglomerative (Hierarchical) Clustering
**Agglomerative Clustering** builds a hierarchy of clusters bottom-up, each image starts as its own cluster and merges progressively.  
Unlike K-Means, it does **not** require random initialisation and often finds more natural groupings.

We evaluate silhouette scores for k = 2 to 15 here as well.



```python
# ── Agglomerative Clustering — Silhouette Evaluation ─────────────────────────
agg_sil_scores = []

for k in k_range:
    agg = AgglomerativeClustering(n_clusters=k, linkage='ward')
    labels = agg.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    agg_sil_scores.append(score)
    print(f"  k={k:2d}  silhouette score={score:.4f}")

best_k_agg = list(k_range)[np.argmax(agg_sil_scores)]
print(f"\n Best k for Agglomerative Clustering (highest silhouette): k = {best_k_agg}")
```

```python
best_k_agg = list(k_range)[np.argmax(agg_sil_scores)]

plt.figure()
plt.plot(list(k_range), agg_sil_scores, marker='o')

plt.axvline(best_k_agg, linestyle='--')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Agglomerative Clustering Silhouette Scores")

plt.show()
```

---
## Cell 7 (a) : Comparing Both Methods Side-by-Side
We plot both silhouette curves on the same axes to visually compare which method is more confident about the cluster structure.

```python
# ── Side-by-side comparison plot ──────────────────────────────────────────────
plt.figure(figsize=(10, 5))
plt.plot(list(k_range), kmeans_sil_scores, 'rs-', linewidth=2, label='K-Means')
plt.plot(list(k_range), agg_sil_scores,    'b^-', linewidth=2, label='Agglomerative (Ward)')
plt.xlabel('Number of Clusters (k)', fontsize=13)
plt.ylabel('Silhouette Score', fontsize=13)
plt.title('Silhouette Score Comparison: K-Means vs Agglomerative', fontsize=14)
plt.xticks(list(k_range))
plt.legend(fontsize=12)
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()

# ── Summary table ─────────────────────────────────────────────────────────────
results_df = pd.DataFrame({
    'k': list(k_range),
    'KMeans_Silhouette': [round(s, 4) for s in kmeans_sil_scores],
    'Agglomerative_Silhouette': [round(s, 4) for s in agg_sil_scores]
})
print(results_df.to_string(index=False))
print(f"\nK-Means best k      : {best_k_kmeans}")
print(f" Agglomerative best k: {best_k_agg}")

# ── Choose the CORRECT number of clusters ─────────────────────────────────────
# Based on domain knowledge (social media image categories) and evaluation above,
# we set the final number of clusters. Adjust this to match your elbow/silhouette findings.
OPTIMAL_K = best_k_kmeans   # <── Change this value if your visual inspection suggests differently
print(f"\n Using OPTIMAL_K = {OPTIMAL_K} for tasks b, c, and d.")
```

---
## Cell 8 (a) : Train Final K-Means Model with Optimal K
We now fit the **final** K-Means model using the chosen optimal number of clusters.

```python
# ── Train final K-Means with the chosen OPTIMAL_K ─────────────────────────────
final_kmeans = KMeans(n_clusters=OPTIMAL_K, random_state=42, n_init=10)
cluster_labels = final_kmeans.fit_predict(X_pca)

# Check how many images landed in each cluster
unique, counts = np.unique(cluster_labels, return_counts=True)
print("Cluster label → Image count:")
for label, count in zip(unique, counts):
    print(f"  Cluster {label:2d}: {count} images")

final_silhouette = silhouette_score(X_pca, cluster_labels)
print(f"\nFinal K-Means silhouette score (k={OPTIMAL_K}): {final_silhouette:.4f}")
```

---
## Cell 9 (b) : Build the Cluster Dictionary
A dictionary where:
- **Key** = cluster label (integer)
- **Value** = list of image filenames assigned to that cluster

This is the core data structure for all downstream tasks.

```python
# ── Build cluster → [filenames] dictionary ────────────────────────────────────
cluster_dict = {}   # {cluster_label: [filename1, filename2, ...]}

for filename, label in zip(filenames, cluster_labels):
    if label not in cluster_dict:
        cluster_dict[label] = []       # Initialise empty list on first encounter
    cluster_dict[label].append(filename)

# ── Print a preview ───────────────────────────────────────────────────────────
print("Cluster Dictionary Preview (first 3 filenames per cluster):")
print("-" * 60)
for label in sorted(cluster_dict.keys()):
    preview = cluster_dict[label][:3]  # Show only first 3 filenames for brevity
    print(f"Cluster {label:2d} ({len(cluster_dict[label])} images): {preview}")
```

---
## Cell 10 (c) : Move Images into Cluster Folders
We create a folder per cluster and **copy** (or move) each image into the correct subfolder.  
Using `shutil.copy` instead of `shutil.move` so the originals in `collected_images/` are preserved.

```python
# ── Configuration ─────────────────────────────────────────────────────────────
SOURCE_FOLDER = 'collected_images'       # Where the original images live
OUTPUT_FOLDER = 'clustered_images'       # Parent folder for cluster sub-folders

# Create the parent output folder if it doesn't already exist
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# ── Copy each image to its cluster subfolder ──────────────────────────────────
for label, img_list in cluster_dict.items():
    cluster_folder = os.path.join(OUTPUT_FOLDER, f'cluster_{label}')
    os.makedirs(cluster_folder, exist_ok=True)   # Create cluster subfolder

    for img_name in img_list:
        src = os.path.join(SOURCE_FOLDER, img_name)
        dst = os.path.join(cluster_folder, img_name)
        if os.path.exists(src):
            shutil.copy(src, dst)                # Copy the file

print(f"Images organised into '{OUTPUT_FOLDER}/' with one sub-folder per cluster.")
print("Sub-folders created:")
for folder in sorted(os.listdir(OUTPUT_FOLDER)):
    n = len(os.listdir(os.path.join(OUTPUT_FOLDER, folder)))
    print(f"  {folder}/  →  {n} images")
```

---
## Cell 11 (c) : Display 5 Random Images from Each Cluster
We display a small gallery for each cluster to visually verify that images in the same cluster look similar.

```python
# ── Show 5 random images from every cluster ───────────────────────────────────
IMAGES_TO_SHOW = 5

for label in sorted(cluster_dict.keys()):
    img_list = cluster_dict[label]
    # If a cluster has fewer than 5 images, show all of them
    sample = random.sample(img_list, min(IMAGES_TO_SHOW, len(img_list)))

    fig, axes = plt.subplots(1, len(sample), figsize=(4 * len(sample), 4))
    fig.suptitle(f'Cluster {label}  ({len(img_list)} images total)', fontsize=14, fontweight='bold')

    # Handle the edge case where only 1 image is in the sample (axes won't be a list)
    if len(sample) == 1:
        axes = [axes]

    for ax, img_name in zip(axes, sample):
        img_path = os.path.join(SOURCE_FOLDER, img_name)
        if os.path.exists(img_path):
            img = mpimg.imread(img_path)
            ax.imshow(img)
        else:
            ax.text(0.5, 0.5, 'Not found', ha='center')  # Graceful fallback
        ax.set_title(img_name, fontsize=7)
        ax.axis('off')

    plt.tight_layout()
    plt.show()
```

---
## Cell 12 (d) : Identifying Duplicate Images Using Cosine Distance
### Why Cosine Distance?
**Cosine distance** measures the angle between two vectors, ignoring magnitude.  
If two images are identical (or near-identical), their embedding vectors point in almost the same direction → **cosine distance ≈ 0**.

We set a small threshold (e.g., 0.01) — any pair with distance below this threshold is flagged as a **duplicate**.

```python
# ── Compute pairwise cosine distances ─────────────────────────────────────────
# cosine_distances returns a (N x N) matrix; entry [i,j] = distance between image i and j
print("Computing pairwise cosine distances (this may take a moment)...")
dist_matrix = cosine_distances(X)   # Use original 2048-D vectors for accuracy
print(f"Distance matrix shape: {dist_matrix.shape}")

# ── Find duplicate pairs ──────────────────────────────────────────────────────
DUPLICATE_THRESHOLD = 0.01  # Images with cosine distance < threshold are duplicates

duplicates = []  # Will hold tuples of (filename_A, filename_B, distance)
n = len(filenames)

# Only iterate the upper triangle (i < j) to avoid counting each pair twice
for i in range(n):
    for j in range(i + 1, n):
        if dist_matrix[i, j] < DUPLICATE_THRESHOLD:
            duplicates.append((filenames[i], filenames[j], dist_matrix[i, j]))

print(f"\nDuplicate threshold : cosine distance < {DUPLICATE_THRESHOLD}")
print(f"Number of duplicate pairs found: {len(duplicates)}")
```

---
## Cell 13 (d) : Display Duplicate Pairs
Print a tidy table of all detected duplicate image pairs, sorted from most-similar to least.

```python
# ── Show duplicate pairs as a sorted DataFrame ────────────────────────────────
if duplicates:
    dup_df = pd.DataFrame(duplicates, columns=['Image_A', 'Image_B', 'Cosine_Distance'])
    dup_df = dup_df.sort_values('Cosine_Distance')   # Most similar pairs first
    dup_df['Cosine_Distance'] = dup_df['Cosine_Distance'].round(6)
    print(dup_df.to_string(index=False))
else:
    print("No exact or near-exact duplicates found at the current threshold.")
    print("Try increasing DUPLICATE_THRESHOLD (e.g., 0.05) to find near-duplicates.")
```

---
## Cell 14 (d) : Visualise Duplicate Pairs Side-by-Side
Display each duplicate pair as a side-by-side figure so we can visually confirm the finding.

```python
# ── Plot up to 5 duplicate pairs side-by-side ─────────────────────────────────
MAX_DISPLAY = 5

if duplicates:
    for img_a, img_b, dist in duplicates[:MAX_DISPLAY]:
        path_a = os.path.join(SOURCE_FOLDER, img_a)
        path_b = os.path.join(SOURCE_FOLDER, img_b)

        if os.path.exists(path_a) and os.path.exists(path_b):
            fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))
            fig.suptitle(f'Duplicate Pair  |  Cosine Distance = {dist:.6f}', fontsize=12)

            ax1.imshow(mpimg.imread(path_a))
            ax1.set_title(img_a, fontsize=8)
            ax1.axis('off')

            ax2.imshow(mpimg.imread(path_b))
            ax2.set_title(img_b, fontsize=8)
            ax2.axis('off')

            plt.tight_layout()
            plt.show()
else:
    print("No duplicates to display.")
```
